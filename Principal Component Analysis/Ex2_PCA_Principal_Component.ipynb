# Exercise 2 – Identify the Principal Component Variables

The following dataset is taken directly from the question:

| Variable | 1 | 2 | 3 |
|---|---:|---:|---:|
| Climate | 0.190 | 0.017 | 0.207 |
| Housing | 0.544 | 0.020 | 0.204 |
| Health | 0.782 | -0.605 | 0.144 |
| Crime | 0.365 | 0.294 | 0.585 |
| Transportation | 0.585 | 0.085 | 0.234 |
| Education | 0.394 | -0.273 | 0.027 |
| Arts | 0.985 | 0.126 | -0.111 |
| Recreation | 0.520 | 0.402 | 0.519 |
| Economy | 0.142 | 0.150 | 0.239 |

In PCA, the variable with the largest **absolute loading** is the strongest contributor to that component. The sign indicates direction. citeturn0search1turn0search9


```python
import pandas as pd

loadings = pd.DataFrame({
    "Variable": [
        "Climate", "Housing", "Health", "Crime",
        "Transportation", "Education", "Arts",
        "Recreation", "Economy"
    ],
    "PC1": [0.190, 0.544, 0.782, 0.365, 0.585, 0.394, 0.985, 0.520, 0.142],
    "PC2": [0.017, 0.020, -0.605, 0.294, 0.085, -0.273, 0.126, 0.402, 0.150],
    "PC3": [0.207, 0.204, 0.144, 0.585, 0.234, 0.027, -0.111, 0.519, 0.239]
})

loadings
```

## Find the strongest variable for each principal component

```python
for pc in ["PC1", "PC2", "PC3"]:
    idx = loadings[pc].abs().idxmax()
    variable = loadings.loc[idx, "Variable"]
    value = loadings.loc[idx, pc]
    print(f"{pc}: {variable} (loading = {value})")
```

## Find the strongest loading in the entire table

```python
long = loadings.melt(
    id_vars="Variable",
    var_name="Principal_Component",
    value_name="Loading"
)

long["Absolute_Loading"] = long["Loading"].abs()

strongest = long.loc[long["Absolute_Loading"].idxmax()]

print("Variable:", strongest["Variable"])
print("Principal Component:", strongest["Principal_Component"])
print("Loading:", strongest["Loading"])
print("Absolute loading:", strongest["Absolute_Loading"])
```

## Answer

- **PC1 → Arts**, loading = **0.985**
- **PC2 → Health**, loading = **-0.605**
- **PC3 → Crime**, loading = **0.585**

Therefore, if the question asks for the **single variable with the strongest loading in the whole table**, the answer is:

### **Arts — PC1 (0.985)**

The negative sign for Health on PC2 means an inverse relationship with that component; for identifying the strongest contribution, we compare absolute loading values. citeturn0search1turn0search9
